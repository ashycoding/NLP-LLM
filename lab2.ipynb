```
import numpy as np
import pandas as pd

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Sample corpus
corpus = [
    "the quick brown fox jumps over the lazy dog",
    "the lazy dog barks at the quick brown fox",
    "the fox and the dog are friends"
]

print("Sample Corpus:")
for i, doc in enumerate(corpus):
    print(f"Document {i+1}: {doc}")
```

```
#TF IDF
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(corpus)

df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print("TF-IDF:")
print(df.round(2))
```

```
#Co Occurence Matrix

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)

# Word-document matrix
X = X.toarray()

# Word-word co-occurrence matrix
co_matrix = X.T @ X

words = vectorizer.get_feature_names_out()

df = pd.DataFrame(
    co_matrix,
    index=words,
    columns=words
)

print("Co-occurrence Matrix:")
print(df)
```

```
#Word2Vec

from gensim.models import Word2Vec

sentences = [sentence.split() for sentence in corpus]

model = Word2Vec(
    sentences,
    vector_size=50,
    window=2,
    min_count=1,
    workers=1
)

print("Vector for 'fox':")
print(model.wv["fox"])

print("\nSimilar words to 'fox':")
print(model.wv.most_similar("fox"))
```

```
import requests
import zipfile
import os

url = "https://nlp.stanford.edu/data/glove.6B.zip"

r = requests.get(url)
open("glove.6B.zip", "wb").write(r.content)

with zipfile.ZipFile("glove.6B.zip", "r") as zip_ref:
    zip_ref.extractall(".")

print("GloVe downloaded!")
```

```
embeddings = {}

with open("glove.6B.50d.txt", encoding="utf8") as f:
    for line in f:
        values = line.split()
        word = values[0]
        vector = np.array(values[1:], dtype=float)
        embeddings[word] = vector

print("Vector for fox:")
print(embeddings["fox"])
```
